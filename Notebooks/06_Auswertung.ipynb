## Evaluationsberichte

Für jede der folgenden Modellkonfigurationen wird ein Evaluationsbericht
erzeugt. Die Baseline-Konfiguration verwendet den vollständigen Datensatz
mit Nebendiagnosen, Kovariaten und ICD-Codes als Input. Alle weiteren
Konfigurationen variieren jeweils einen Parameter gegenüber der Baseline,
um dessen Einfluss isoliert beurteilen zu können.

| Konfiguration | `use_sdx` | `subset` | `kovarianten` | `input_var` |
|---|---|---|---|---|
| Baseline | ✓ | gesamt | ✓ | codes |
| Ohne Nebendiagnosen | ✗ | gesamt | ✓ | codes |
| Nur medizinische Fälle | ✓ | partition=M | ✓ | codes |
| Nur operative Fälle | ✓ | partition=O | ✓ | codes |
| Nur Notfälle | ✓ | aufn_anl=N | ✓ | codes |
| Ohne Notfälle | ✓ | aufn_anl=E,V,G | ✓ | codes |
| Ohne Kovariaten | ✓ | gesamt | ✗ | codes |
| Charlsonlike-Score | ✓ | gesamt | ✓ | charlson |
| Elixhauserlike-Score | ✓ | gesamt | ✓ | elixhauser |

```python
# Baseline
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = "gesamt", kovarianten = True, input_var = "codes")

# Keine Nebendiagnosen
typst_table_exporter.generate_evaluation_reports(use_sdx = False, subset = "gesamt", kovarianten = True, input_var = "codes")

# nur medizinische Fälle
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = {"partition": "M"}, kovarianten = True, input_var = "codes")

# nur operative Fälle
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = {"partition": "O"}, kovarianten = True, input_var = "codes")

# nur Notfälle
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = {"aufn_anl": "N"}, kovarianten = True, input_var = "codes")

# keine Notfälle
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = {"aufn_anl": "E,V,G"}, kovarianten = True, input_var = "codes")

# keine Kovarianten
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = "gesamt", kovarianten = False, input_var = "codes")

# Charlsonlike Score
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = "gesamt", kovarianten = True, input_var = "charlson")

# Elixhauser Score
typst_table_exporter.generate_evaluation_reports(use_sdx = True, subset = "gesamt", kovarianten = True, input_var = "elixhauser")
```

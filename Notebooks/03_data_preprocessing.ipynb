# Aufbereitung der Campus File-Daten

## Zweck

Dieses Notebook bereinigt und validiert die Rohdaten des DRG-Campus Files 2010.
Die Verarbeitungslogik ist vollständig in der Klasse `DataPreprocessor` gekapselt;
dieses Notebook dient als Einstiegspunkt zur Ausführung.

Folgende Schritte werden durchgeführt (ausführliche Erläuterung in der Arbeit):

- Datenbereinigung und Plausibilitätsprüfung der Campus File-Rohdaten
- Aggregation der ICD-Codes von der Gruppen- auf die Kapitel-Ebene
- Export als `.feather`-Dateien für schnelleres Einlesen in nachfolgende Notebooks

## Eingabe

| Datei | Beschreibung |
|---|---|
| `Campus_File_DRG_2010_DRG_Statistik.csv` | Rohdaten des DRG-Campus Files |
| `icd_systematik_2010.json` | ICD-10-Hierarchie (Notebook 1) |
| `ops_systematik_2010.json` | OPS-Hierarchie |
| `columns_metadata.json` | Spaltenbeschreibungen und Datentypen |
| `icd_subgruppen.json` | ICD-Subgruppen für die Aggregation |
| `Fallpauschalenkatalog_2010.xls` | DRG-Fallpauschalenkatalog 2010 |

## Ausgabe

- `.feather`-Dateien der bereinigten Daten (Pfade durch `DataPreprocessor` verwaltet)

```python
from drg_risk.data.datapreprocessor import DataPreprocessor
from drg_risk import DATA_DIR
import os
```

```python
preprocessor = DataPreprocessor(
    data_path =                 os.path.join(DATA_DIR, "Campus_File_DRG_2010", "Campus_File_DRG_2010_DRG_Statistik.csv"),
    icd_sys_path =              os.path.join(DATA_DIR, "icd_systematik_2010.json"),
    ops_sys_path =              os.path.join(DATA_DIR, "ops_systematik_2010.json"),
    columns_metadata_path =     os.path.join(DATA_DIR, "Campus_File_DRG_2010", "columns_metadata.json"),
    subgroups_path =            os.path.join(DATA_DIR, "icd_subgruppen.json"),
    drg_path =                  os.path.join(DATA_DIR, "Fallpauschalenkatalog_2010.xls"),
)

preprocessor.process_data()
```
